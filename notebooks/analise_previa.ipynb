```python
import pandas as pd
```

```python
url = r'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv'

df = pd.read_csv(url)
display(df.head())
display(df.tail())
```

```python
# Passando métricas de Análise

display('Valores Duplicados:', df[df.duplicated(keep=False)])
display('Valores nulos:', df.isnull().sum())
display('Linhas x Colunas:', df.shape)
display('Linhas x Colunas:', df.count())
display('Tipos de Dados:', df.dtypes)
display('Estatísticas de dados:', df.describe())

display(f'Antes: {len(df)} valores')
```

```python
# Pós remoção de duplicata
df = df.drop_duplicates()

df[df.duplicated(keep=False)]
print(f'Após remoção: {len(df)} valores')
```
