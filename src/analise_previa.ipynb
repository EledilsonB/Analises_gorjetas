```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
url = r'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv'

df = pd.read_csv(url)
df.head()
```

```python
# Passando métricas de Análise

# duplicata = [dup for dup in df.duplicated(keep=False) if dup == True]

display('Valores Duplicados:', df[df.duplicated(keep=False)])
display('Valores nulos:', df.isnull().sum())
display('Linhas x Colunas:', df.count())
display('Tipos de Dados:', df.dtypes)
display('Estatísticas de dados:', df.describe())

display(f'Antes: {len(df)} valores')
```

```python
# Pós remoção de duplicata
df = df.drop_duplicates()

df[df.duplicated(keep=False)]
print(f'Após remoção: {len(df)} valores')
```

```python
# Repassando distribuição

display('Estatísticas de dados:', df.describe())
```

### Colinha de Análise feitas pelo Describe:
| `describe()` | O que significa                                    |
| ------------ | -------------------------------------------------- |
| **count**    | Quantidade de valores **não nulos**                |
| **mean**     | Média aritmética                                   |
| **std**      | Desvio-padrão; indica a **dispersão** dos dados    |
| **min**      | Menor valor                                        |
| **25%**      | 1º quartil (Q1): 25% dos valores estão abaixo dele |
| **50%**      | Mediana (Q2): 50% dos valores estão abaixo dele    |
| **75%**      | 3º quartil (Q3): 75% dos valores estão abaixo dele |
| **max**      | Maior valor                                        |
